```python
from numpy import arange
```

```python
def function(x,y):
  return x**2 + y**2

r_max =5
r_min = -5
step= 0.1
sample=[]

for x in arange(r_min,r_max,step):
  for y in arange(r_min,r_max,step):
    sample.append([x,y])

best_eval=1000000000000000000
x_best,y_best=0 , 0

for x,y in sample :
  eval = function(x,y)
  if eval< best_eval :
    x_best= x
    y_best = y
    best_eval = eval

print(x_best,y_best, best_eval)

```
